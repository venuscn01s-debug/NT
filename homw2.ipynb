```python
import math
a,b,c=map(int,input().split())
f=(a+b+c)/2
k=f*(f-a)*(f-b)*(f-c)
if k>=0 and a+b>c and a+c>b and b+c>a:
    d=math.sqrt(f*(f-a)*(f-b)*(f-c))
    if a==b==c:
        print(f'tgd có diện tích :{d} ,chu vi{f*2}')
    elif a==b!=c and a==c!=b and b==c!=a:
        print(f'tgc có diện tích :{d} ,chu vi{f*2}')
    elif a**2+b**2==c**2 or a**2+c**2==b**2 or c**2+b**2==a**2:
        print(f'tgv có diện tích :{d} ,chu vi{f*2}')
    else:
        print(f'tgt có diện tích :{d} ,chu vi{f*2}')
else:
    print('kp tg')
```

```python
def doc(n):
    cs=["không","một","hai","ba","bốn","năm","sáu","bảy","tám","chín"]; dv=["","nghìn","triệu","tỷ"]
    def d(x,f):
        t,c,dv1=x//100,(x%100)//10,x%10
        s=(cs[t]+" trăm" if t else ("không trăm" if f and (c or dv1) else ""))
        s+=(" "+cs[c]+" mươi" if c>1 else " mười" if c==1 else "")
        s+=(" mốt" if dv1==1 and c>1 else " lăm" if dv1==5 and c else (" lẻ "+cs[dv1] if dv1 and not c else (" "+cs[dv1] if dv1 else "")))
        return s.strip()
    kq,i,f="",0,1
    while n: b=n%1000; kq=(d(b,not f)+" "+dv[i]+" "+kq if b else kq); f=0; n//=1000; i+=1
    return kq.strip() or "không"

print(doc(int(input())))
```
