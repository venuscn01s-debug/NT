```python
n = int(input())

max_len = 0
result = 1

x = 1
while x <= n:
    temp = x
    length = 1

    while temp != 1:
        if temp % 2 == 0:
            temp = temp // 2
        else:
            temp = temp * 3 + 1
        length += 1

    if length > max_len:
        max_len = length
        result = x

    x += 1

print(result, max_len)
```
