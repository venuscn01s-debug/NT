```python
a=int(input())
b=abs(a)
if a>0 :
    print(f'{str(a)[::-1]}')
elif a==0 :
    print('0')
else:
    print(f'-{str(b)[::-1]}')
```

```python
arr=list(map(int,input().split()))
arr.sort(reverse=True)
max=arr[0]
for i in arr :
    if i!=max:
        print(i)
        break
```
