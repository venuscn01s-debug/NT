```python
gia={1:1000,2:2000,3:3000,4:4000}
tong=0
while True :
    masp=int(input())
    if masp==0:
        break
    x=int(input())
    tong=tong+x*gia[masp]
print(tong)
```

```python
gia={1:1000,2:2000,3:3000,4:4000}
n=input()
ds=n.split()
sum=0
for sp in ds :
    ma,sl=sp.split('-')
    ma=int(ma)
    sl=int(sl)
sum+=gia[ma]*sl
print(sum)
```
