```python
def x(k):
    s=0
    for i in range(1,k):
        if k%i==0:
            s+=i
    return s
n,m=map(int,input().split())
if n==x(m) and m==x(n):
    print("True")
else:
    print('False')
```

```python
def x(lst, k):
    for i in range(len(lst)):
        if lst[i] == k:
            return i
    else:
        return -1
lst=list(map(int,input().split()))
k=int(input())
print(x(lst,k))
```

```python
def x(n):
    s=1
    for i in range(1,n+1):
        s*=i
    return s
n=int(input())
print(x(n))
```

```python
def c(n1,n2,o):
    if o=='+':
        return round(n1+n2,2)
    elif o=='-':
        return round(n1-n2,2)
    elif o=='*':
        return round(n1*n2,2)
    elif o=='/':
        if n2==0:
            return "error"
        else:
            return round(n1/n2,2)
n1=int(input())
n2=int(input())
o=input()
print((c(n1,n2,o)))
```

```python
def x(n):
    s=0
    while n>0:
        x=n%10
        s+=x
        n//=10
    return s
n=int(input())
print(x(n))
```

```python
def x(a, b):
    if len(a) != len(b):
        return False
    
    m_ab = {}
    m_ba = {}
    
    for i in range(len(a)):
        if a[i] in m_ab and m_ab[a[i]] != b[i]:
            return False
        if b[i] in m_ba and m_ba[b[i]] != a[i]:
            return False
        
        m_ab[a[i]] = b[i]
        m_ba[b[i]] = a[i]
    
    return True
a,b=input().split()
print(x(a, b))
```
