```python
nums = []

while True:
    x = int(input())
    if x == -1:
        break
    nums.append(x)

if nums:
    print("Max:", max(nums))
    print("Min:", min(nums))
else:
    print("Không có số nào")
```

```python
s = input()

co = 0
i = False

for c in s:
    if c != ' ':
        if not i:
            co += 1
            i= True
    else:
        i= False

print(co)
```

```python
s = input()

re = ""
for c in s:
    if 'a' <= c <= 'z':
        re += chr(ord(c) - 32)
    elif 'A' <= c <= 'Z':
        re += chr(ord(c) + 32)
    else:
        re += c

print(re)
```
